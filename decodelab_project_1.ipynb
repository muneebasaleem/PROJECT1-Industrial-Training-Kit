**Project Overview**

This project focuses on cleaning and preparing a raw e-commerce dataset using Python and Pandas. The dataset was analyzed for missing values, duplicate records, inconsistent date formats, and incorrect data types. The cleaned dataset is suitable for further analysis and visualization.

**Objectives**

* Handle missing values

* Remove duplicate records

* Standardize date formats

* Validate data types

* Improve overall data quality

**Tools Used**

*  Python
*  Pandas
*  Google Colab


**Step 1: Import Libraries**

```python
import pandas as pd
```

```python
from google.colab import files
uploaded=files.upload()
```

**Step 2: Load Dataset**

```python
df=pd.read_csv("Dataset for Data Analytics.csv")
```

**Step 3: Dataset Overview**

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

df.head()
```

**Step 4: Missing Value Analysis and Handling**

```python
print("Missing Values Before Cleaning:")
print(df.isnull().sum())

df["CouponCode"] = df["CouponCode"].fillna("No Coupon")

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

**Step 5: Duplicate Record Detection and Removal**

```python
print("Duplicate Rows Before Cleaning:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())

print("\nDuplicate Order IDs:")
print(df["OrderID"].duplicated().sum())
```

**Step 6:Date Format Standardization**

```python
# Convert the "Date" column to datetime format (if it's not already)
df["Date"] = pd.to_datetime(df["Date"])

# Format the datetime values into 'YYYY-MM-DD' string format
df["Date"] = df["Date"].dt.strftime("%Y-%m-%d")

# Display the first 5 values of the updated "Date" column
print(df["Date"].head())
```

**Step 7: Data Type Validation**

```python
# Convert "Quantity" column to numeric (handles integers or floats)
df["Quantity"] = pd.to_numeric(df["Quantity"])

# Convert "UnitPrice" column to numeric
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"])

# Convert "TotalPrice" column to numeric
df["TotalPrice"] = pd.to_numeric(df["TotalPrice"])

# Print data types of all columns to verify conversions
print(df.dtypes)
```

**Step 8: Final Data Quality Check**

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDuplicate Order IDs:")
print(df["OrderID"].duplicated().sum())
```

**Step 8: Export Cleaned Dataset**

```python
df.to_csv("Cleaned_Dataset.csv", index=False)

print("Project Completed Successfully!")
```
